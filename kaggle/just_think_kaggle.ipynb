# Just Think on Kaggle

This runs the main study (Qwen 2.5 7B Instruct, 1,700 trials), unattended. Set the accelerator to **GPU T4 x2**, turn **Internet** on, then use **Save Version > Save & Run All (Commit)** and close the tab. Everything runs on Kaggle's servers. Results appear in this version's **Output** tab under `results/`.

If a session ends before everything is done, open the notebook again, attach the previous version's output with **Add Input > Your Work**, and commit again. It carries on from the next missing trial.

```python
import os, subprocess
REPO_URL = "https://github.com/pragyaangaur/Just-Think"
CODE = "/tmp/just-think"
if not os.path.exists(CODE):
    subprocess.run(["git", "clone", REPO_URL, CODE], check=True)
    subprocess.run(["git", "clone", "https://github.com/valen-research/Pain-axis", f"{CODE}/external/Pain-axis"], check=True)
    subprocess.run(["git", "-C", f"{CODE}/external/Pain-axis", "checkout", "4d75cd90e206ea962f7a9101e65c85efea56723b"], check=True)
print("code at", subprocess.run(["git", "-C", CODE, "log", "-1", "--format=%H %s"], capture_output=True, text=True).stdout)
```
